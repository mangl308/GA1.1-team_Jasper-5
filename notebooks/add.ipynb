```python
def add(a, b):
    '''This function adds two numbers as input and returns their sum'''
    addition = a + b
    return addition
```

```python
def sum_list(values):
    total = 0
    for i in values:
        total = add(total, i)
    return total
```

```python
values = [1, 2, 3, 14, 5]
sum_list(values)
```
